```python
import re

# Extract the first coordinate pair from each MULTIPOLYGON as a centroid approximation
def extract_centroid(geom):
    coords = re.findall(r'(-?\d+\.\d+)\s(-?\d+\.\d+)', geom)
    if coords:
        lons = [float(c[0]) for c in coords]
        lats = [float(c[1]) for c in coords]
        return sum(lats)/len(lats), sum(lons)/len(lons)
    return None, None

nta['centroid_lat'], nta['centroid_lon'] = zip(*nta['the_geom'].apply(extract_centroid))

print(nta[['ntaname', 'centroid_lat', 'centroid_lon']].head(10))
```

```python
import re

def extract_bbox(geom):
    # Extract all numbers from the geometry string
    coords = re.findall(r'-?\d+\.\d+', geom)
    # Split into lons and lats (they alternate)
    lons = [float(coords[i]) for i in range(0, len(coords), 2)]
    lats = [float(coords[i]) for i in range(1, len(coords), 2)]
    return min(lons), max(lons), min(lats), max(lats)

# Apply to each neighborhood
nta[['min_lon', 'max_lon', 'min_lat', 'max_lat']] = nta['the_geom'].apply(
    lambda x: pd.Series(extract_bbox(x))
)

print(nta[['ntaname', 'min_lon', 'max_lon', 'min_lat', 'max_lat']].head())
```

```python
def assign_neighborhood(lat, lon):
    match = nta[
        (nta['min_lat'] <= lat) & (lat <= nta['max_lat']) &
        (nta['min_lon'] <= lon) & (lon <= nta['max_lon'])
    ]
    if len(match) > 0:
        return match.iloc[0]['ntaname']
    return None

# Apply to each venue (this may take a minute)
venues['neighborhood'] = venues.apply(
    lambda row: assign_neighborhood(row['latitude'], row['longitude']), axis=1
)

print(venues['neighborhood'].value_counts().head(10))
print("Unmatched venues:", venues['neighborhood'].isna().sum())
```

```python
# Aggregate by neighborhood
neighborhood_scores = venues.groupby(['neighborhood', 'premisescounty']).agg(
    venue_count=('licensepermitid', 'count'),
    avg_subway_entrances=('nearby_subway_entrances', 'mean'),
    avg_complaints=('nearby_complaints', 'mean')
).reset_index()

# Build scores
neighborhood_scores['subway_score'] = (neighborhood_scores['avg_subway_entrances'] / neighborhood_scores['avg_subway_entrances'].max()) * 100
neighborhood_scores['safety_score'] = (1 - neighborhood_scores['avg_complaints'] / neighborhood_scores['avg_complaints'].max()) * 100
neighborhood_scores['competition_score'] = (1 - neighborhood_scores['venue_count'] / neighborhood_scores['venue_count'].max()) * 100

# Final PopupIQ score
neighborhood_scores['popupiq_score'] = (
    neighborhood_scores['subway_score'] * 0.4 +
    neighborhood_scores['safety_score'] * 0.4 +
    neighborhood_scores['competition_score'] * 0.2
).round(1)

# Sort by score
neighborhood_scores = neighborhood_scores.sort_values('popupiq_score', ascending=False).reset_index(drop=True)

print(neighborhood_scores[['neighborhood', 'premisescounty', 'venue_count', 'popupiq_score']].head(20))
```

```python
# Fix 1: Drop rows where borough doesn't match neighborhood's actual borough
borough_map = {
    'Bronx': 'Bronx',
    'Kings': 'Brooklyn',
    'New York': 'Manhattan',
    'Queens': 'Queens',
    'Richmond': 'Staten Island'
}

neighborhood_scores['boro_check'] = neighborhood_scores['premisescounty'].map(borough_map)

# Keep only rows where NTA borough matches
neighborhood_scores = neighborhood_scores.merge(
    nta[['ntaname', 'boroname']],
    left_on='neighborhood',
    right_on='ntaname'
)

neighborhood_scores = neighborhood_scores[
    neighborhood_scores['boroname'] == neighborhood_scores['boro_check']
]

# Fix 2: Only keep neighborhoods with at least 5 venues
neighborhood_scores = neighborhood_scores[neighborhood_scores['venue_count'] >= 5]

# Re-sort
neighborhood_scores = neighborhood_scores.sort_values('popupiq_score', ascending=False).reset_index(drop=True)

print(neighborhood_scores[['neighborhood', 'boroname', 'venue_count', 'popupiq_score']].head(20))
```
